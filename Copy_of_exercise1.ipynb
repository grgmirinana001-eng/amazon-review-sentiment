```python
def nWordTypes(tokenizer, docs: List[str]):
    word_types = set()

    for doc in docs:
        tokens = tokenizer.tokenize(doc)
        word_types.update(tokens)

    return len(word_types)
```

```python
from collections import Counter
import math
import matplotlib.pyplot as plt

# Tokenize all tweets in the dataset
all_tokens = []

for doc in df["text"]:
    all_tokens.extend(ptb_tokenizer.tokenize(doc))

# Count how often each token appears
word_counts = Counter(all_tokens)

# Get words ordered from most frequent to least frequent
most_common = word_counts.most_common()

# Create ranks starting at 1
ranks = list(range(1, len(most_common) + 1))

# Get frequencies
frequencies = [count for word, count in most_common]

# Take log base 10 of rank and frequency
log_ranks = [math.log10(rank) for rank in ranks]
log_frequencies = [math.log10(freq) for freq in frequencies]

# Make the Zipf plot
plt.figure(figsize=(10, 6))
plt.scatter(log_ranks, log_frequencies, s=10)

plt.xlabel("Log Rank (base 10)")
plt.ylabel("Log Frequency (base 10)")
plt.title("Zipf Plot of Word Frequencies")
plt.show()
```

The plot makes sense, we see a clear downward trend. The words with high frequencies have low ranks and the words with higher ranks are less frequent. That's why we would expect from Zipf's law. Only a few words are used very often, and many others are used only a few times. The relationship is easier to see in the log-log plot.
